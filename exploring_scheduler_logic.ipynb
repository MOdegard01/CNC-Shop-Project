```python
from datetime import datetime, date, time, timedelta

#-----------------------------------
#Specifying break times and Holidays

breaks = [                                  
    (time(9,0), time(9,15)),
    (time(13,0), time(13,30)),
    (time(14,0), time(14,15)),
    (time(17,0), time(17,15)),
    (time(19,0), time(19,30)),
    (time(23,0), time(23,15)),
]

holidays = {date(2026,7,15)}

#--------------------------------------------------
#Function for taking a datetime object and picking whether or not that moment goes into the operation times list

def is_open(dt):
    t = dt.time()

    if time(1,0) <= t < time(7,0):
        return False  

    shift_day = dt.date() - timedelta(days=1) if t < time(1,0) else dt.date()

    if shift_day.weekday() >= 5 or shift_day in holidays:
        return False

    for start, end in breaks:
        if start <= t < end:
            return False

    return True

#--------------------------------------------------------
#Creating an empty list of "elapsed timedeltas past July 1st, 2026 at 0000"

ticks = []

#-------------------------------------------------------------
#Declaring an anchor datetime (t) to begin the operations list window and specifying how much time elapses per step

t = datetime(2026, 7, 1, 0, 0)
end = datetime(2026, 7, 22, 0, 0)
step = timedelta(hours=0.01)

#------------------------------------------------------
#Taking each step expressed as a timedelta and using boolean logic to determine if it goes in the operations calendar

while t < end:
    if is_open(t):
        ticks.append(t)
    t += step

#-------------------------------------------------
#Printing ticks to verify. This is a poor visual, but useful in the script logic.

print(ticks)
```
